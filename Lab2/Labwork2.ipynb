```python
import numba
from numba import cuda
import numpy as np

if cuda.is_available():
    print("Numba found a GPU!")
    has_gpu = True
else:
    print("Numba did not find a GPU.")
    has_gpu = False
```

```python
if has_gpu:

    for i, device in enumerate(cuda.list_devices()):
        print(f"Device {i}:")
        print(f"  Name: {device.name}")
        print(f"  ID: {device.id}")
        print(f"  Compute Capability: {device.compute_capability}")

    if cuda.list_devices():
        cuda.select_device(0)
        current_device = cuda.get_current_device()
        print(f"\nSelected Device: {current_device.name} (ID: {current_device.id})")
        print(f"Compute Capability: {current_device.compute_capability}")
else:
    print("No GPU detected.")
```
